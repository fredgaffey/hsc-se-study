# Pseudocode
Pseudocode is a way of expressing the logic of your algorithm in a language that attempts to be as close to English as possible. 

Once you know the syntax of Pseudocode, it is not very difficult to convert Python to Pseudocode, as Pseudocode algorithms follow the same Sequence - Selection - Repetition format, where you tell the computer what to do in a series of statements, using blocks to group statements together.

## Selection

```python
if age < 18:
	print("You are a child")
else:
	print("You are an adult")
```

**Use ELSEIF for nested if statements, similar to Python's `elif`**

```python
if age < 18:
	print("You are a child")
elif age < 70:
	print("You are an adult")
else:
	print("You are an old adult")
```

**Use the CASEWHERE keyword to choose between many different options. Use this for `match` statements or when you have many `if`/`elif` blocks matching against the value of a variable**

```python
def go_left(): ...
def go_right(): ...
def go_up(): ...
def go_down(): ...

command = input()
match command:
	case "go left" : go_left()
	case "go right": go_right()
	case "go up"   : go_up()
	case "go down" : go_down()
	case _: print("Invalid command!")
```

## Repetition

**Repetition is one of the places where Pseudocode differs from Python the most.**

```python
while age < 20:
	age += 1
```

**In Pseudocode, there is also a POST-TEST loop, where the body is run once before the conditoin is checked**

**In Pseudocode, For loops go between a lower bound and an upper bound, stepping by an amount**

**Note: FOR loops don't use an END FOR, they use NEXT and then the variable. This is to prevent confusion when there are lots of nested FOR loops**

```python
for i in range(10):
	print(i)
```

## Breaking and Control Flow within Loops

**Unlike Python, Pseudocode does not allow for the use of `break`s or `continue`s within loops. Instead, a flag must be used to indicate whether or not the loop should continue.**

```python
while True:
	command = input()
	if command == "quit":
		break
```

**Instead of:**

**do this:**

**If you use a `break` in a loop with a condition, you will need to change the condition to be `condition and not should_quit`.**

**If you use a `break` in a FOR loop, you will have to convert it to a while loop and incorporate the flag into the condition**

**To use `continue`, you should put everything after the continue in the ELSE branch of an IF statement**

```python
for i in range(10):
	if i == 4:
		print("Got 4")
		continue
	print(i)
```

## Arrays

**Arrays in Pseudocode**

**Like in Python, array elements are valid targets for assignment and access**

**In order to loop through an array, you need to use a for loop that goes from the start index to the final index**

```python
names = ["Matthew", "Sarah", "John", "Max", "Mia", "Chris"]

for name in names:
	print("Hello " + name + "!")
```

**Note that you use the notation Array(Index), not `array[index]` like in Python.**

## Records

**You do not need to document the structure of your records in the Pseudocode, however you will need to document them in your Data Dictionary**<br>**In Pseudocode, you can use record values like Python object**

```python
from dataclasses import dataclass

@dataclass
class Person:
	name: str
	age: int

person = Person("John", 32)
person.name = "Alex"
person.age = 24
```

## Some algorithms

**Find max**

```python
def max(array: list[int]) -> int:
	current_max = array[0]
	for elem in array:
		if elem > current_max:
			current_max = elem
	return elem
```

```python
def swap(array: list[int], i: int, j: int): ...

def sort(array: list[int]):
	last: int = len(array)
	swapped: bool = True
	while swapped:
		swapped = False
		i = 1
		while i < last:
			if array[i-1] > array[i]:
				swap(array, i-1, i)
				swapped = True
			i += 1
		last -= 1
```
